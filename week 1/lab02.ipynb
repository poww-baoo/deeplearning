## Môi trường thực thi

```python
import platform
import numpy as np
import torch

SEED = 2026
np.random.seed(SEED)
torch.manual_seed(SEED)

print("Python:", platform.python_version())
print("PyTorch:", torch.__version__)
print("NumPy:", np.__version__)
print("Thiết bị:", "cuda" if torch.cuda.is_available() else "cpu")
print("Seed:", SEED)
```

## Bài 1 Tạo Tensor cơ bản

```python
a = torch.tensor(
    [[1, 2, 3],
     [4, 5, 6]],
    dtype=torch.int32,
)

print(a)
print("Kích thước:", a.shape)
print("Số chiều:", a.ndim)
print("Kiểu dữ liệu:", a.dtype)
```

### Nhận xét và trả lời

Tensor `a` có kích thước `2 × 3`, gồm 2 chiều và có kiểu dữ liệu `torch.int32`.

## Bài 2 Tạo các Tensor thường dùng

```python
torch.manual_seed(SEED)

a_rand = torch.rand(3, 4)
b_randn = torch.randn(3, 4)
c_zeros = torch.zeros(2, 3)
d_ones = torch.ones(2, 3)
e_eye = torch.eye(4)
f_tens = torch.full((5, 5), 10)

print("torch.rand(3, 4):\n", a_rand)
print("\ntorch.randn(3, 4):\n", b_randn)
print("\ntorch.zeros(2, 3):\n", c_zeros)
print("\ntorch.ones(2, 3):\n", d_ones)
print("\ntorch.eye(4):\n", e_eye)
print("\nTensor 5 x 5 toàn giá trị 10:\n", f_tens)
```

### Nhận xét và trả lời

- `torch.rand()` sinh các giá trị theo phân phối đều trên khoảng `[0, 1)`.
- `torch.randn()` sinh các giá trị theo phân phối chuẩn có trung bình 0 và độ lệch chuẩn 1, vì vậy kết quả có thể âm hoặc dương.
- Tensor `5 × 5` chứa toàn giá trị 10 được tạo bằng `torch.full((5, 5), 10)`.

## Bài 3 Kiểm tra thông tin Tensor

```python
a = torch.zeros(2, 3, 4)

print("Tensor:\n", a)
print("Shape:", a.shape)
print("Size:", a.size())
print("Number of dimensions:", a.ndim)
print("Data type:", a.dtype)
```

### Nhận xét và trả lời

`torch.Size([2, 3, 4])` cho biết đây là Tensor 3 chiều. Chiều thứ nhất có 2 khối, mỗi khối có 3 hàng và mỗi hàng có 4 phần tử. Tổng số phần tử là `2 × 3 × 4 = 24`.

## Bài 4 Slicing và Indexing

```python
torch.manual_seed(SEED + 4)
a = torch.randn(3, 4, 5)

print("Original tensor:\n", a)
print("\nTensor a[1]:\n", a[1])
print("Shape a[1]:", a[1].shape)
print("\nTensor a[1:, 2:4]:\n", a[1:, 2:4])
print("Shape a[1:, 2:4]:", a[1:, 2:4].shape)
print("\nHai phần tử cuối trên chiều thứ ba a[:, :, -2:]:\n", a[:, :, -2:])
print("Shape:", a[:, :, -2:].shape)
```

### Nhận xét và trả lời

- `a[1]` cố định chỉ số ở chiều đầu nên bỏ chiều đó; kết quả có shape `(4, 5)`.
- `a[1:, 2:4]` lấy 2 khối cuối ở chiều thứ nhất và 2 hàng có chỉ số 2, 3 ở chiều thứ hai; kết quả có shape `(2, 2, 5)`.
- Hai phần tử cuối trên chiều thứ ba được lấy bằng `a[:, :, -2:]`, có shape `(3, 4, 2)`.

## Bài 5 Reshape và Flatten

```python
torch.manual_seed(SEED + 5)
a = torch.randn(3, 4)
flat = a.ravel()
reshaped_3_2_2 = a.reshape(3, 2, 2)
reshaped_2_6 = a.reshape(2, 6)

print("Original:\n", a)
print("Số phần tử ban đầu:", a.numel())
print("\nFlatten:\n", flat)
print("\nReshape 3 x 2 x 2:\n", reshaped_3_2_2)
print("Số phần tử sau reshape:", reshaped_3_2_2.numel())
print("\nReshape 2 x 6:\n", reshaped_2_6)
print("Shape:", reshaped_2_6.shape)
```

### Nhận xét và trả lời

Tensor ban đầu, Tensor `3 × 2 × 2` và Tensor `2 × 6` đều có 12 phần tử. `reshape()` chỉ thay đổi cách tổ chức các phần tử, không thay đổi dữ liệu hay tổng số phần tử.

## Bài 6 Phép toán và nhân ma trận

```python
torch.manual_seed(SEED + 6)
a = torch.randn(2, 3)
b = torch.randn(2, 3)

print("a:\n", a)
print("\nb:\n", b)
print("\na + b:\n", a + b)
print("\na * b:\n", a * b)
print("\na / b:\n", a / b)
print("\na ** 2:\n", a ** 2)
print("\na @ b.T:\n", a @ b.T)
print("Shape của a @ b.T:", (a @ b.T).shape)
```

### Nhận xét và trả lời

`a * b` nhân từng phần tử ở cùng vị trí nên kết quả giữ shape `(2, 3)`. `a @ b.T` là phép nhân ma trận giữa ma trận `(2, 3)` và `(3, 2)`, vì vậy kết quả có shape `(2, 2)`.

## Bài 7 Các hàm thống kê trên Tensor

```python
torch.manual_seed(SEED + 7)
a = torch.randn(3, 4)

print("Tensor a:\n", a)
print("\nTheo cột dim=0")
print("Mean:", torch.mean(a, dim=0))
print("Standard deviation:", torch.std(a, dim=0))
print("Cumulative sum:\n", torch.cumsum(a, dim=0))

print("\nTheo hàng dim=1")
print("Mean:", torch.mean(a, dim=1))
print("Standard deviation:", torch.std(a, dim=1))
print("Cumulative sum:\n", torch.cumsum(a, dim=1))
```

### Nhận xét và trả lời

Với `dim=0`, PyTorch gộp theo chiều hàng và trả về một giá trị cho mỗi cột, nên mean và std có 4 phần tử; cumulative sum tích lũy từ trên xuống. Với `dim=1`, PyTorch gộp theo chiều cột và trả về một giá trị cho mỗi hàng, nên mean và std có 3 phần tử; cumulative sum tích lũy từ trái sang phải.

## Bài 8 Autograd cơ bản

```python
x = torch.tensor(3.6, requires_grad=True)
y = x * x
y.backward()

print("Với y = x^2")
print("x =", x.item())
print("y =", y.item())
print("x.grad =", x.grad.item())
print("Đạo hàm thủ công 2x =", 2 * x.item())

x2 = torch.tensor(3.6, requires_grad=True)
y2 = 3 * x2**2 + 2 * x2 + 1
y2.backward()

print("\nVới y = 3x^2 + 2x + 1")
print("y =", y2.item())
print("x.grad =", x2.grad.item())
print("Đạo hàm thủ công 6x + 2 =", 6 * x2.item() + 2)
```

### Nhận xét và trả lời

- Với `y = x²`, đạo hàm là `dy/dx = 2x`. Tại `x = 3.6`, kết quả thủ công là `7.2`, trùng với `x.grad` (chỉ có sai số biểu diễn số thực rất nhỏ nếu xuất hiện).
- Với `y = 3x² + 2x + 1`, đạo hàm là `dy/dx = 6x + 2`. Tại `x = 3.6`, gradient bằng `23.6`, khớp phép tính thủ công.

## Bài 9 Gradient Descent tìm hệ số đa thức

```python
np.random.seed(SEED + 9)
torch.manual_seed(SEED + 9)

polynomial = np.poly1d([1, 2, 3])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)
XX = np.hstack([X * X, X, np.ones_like(X)])

x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)
w = torch.randn(3, 1, requires_grad=True)
optimizer = torch.optim.NAdam([w], lr=0.01)

print("Hệ số ban đầu:", w.detach().flatten().tolist())
for epoch in range(2000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()
    if epoch + 1 in (1, 100, 500, 1000, 2000):
        print(f"Epoch {epoch + 1:4d} - MSE = {mse.item():.10f}")

learned = w.detach().flatten()
target = torch.tensor([1.0, 2.0, 3.0])
print("Hệ số học được:", learned.tolist())
print("Hệ số mục tiêu:", target.tolist())
print("Sai lệch tuyệt đối:", torch.abs(learned - target).tolist())
```

### Nhận xét và trả lời

Ba hệ số học được hội tụ rất gần `[1, 2, 3]`; MSE giảm về gần 0 vì dữ liệu được sinh đúng từ đa thức bậc hai và không có nhiễu.

- `optimizer.zero_grad()` xóa gradient của vòng lặp trước vì PyTorch mặc định cộng dồn gradient.
- `mse.backward()` lan truyền ngược trên đồ thị tính toán để tính gradient của MSE theo `w`.
- `optimizer.step()` dùng gradient hiện tại để cập nhật `w` theo thuật toán NAdam.

## Bài 10 Giải hệ phương trình bằng Autograd

```python
torch.manual_seed(SEED + 10)
A = torch.rand((), requires_grad=True)
B = torch.rand((), requires_grad=True)
C = torch.rand((), requires_grad=True)
D = torch.rand((), requires_grad=True)

optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)
for epoch in range(3000):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    sqerr = y1**2 + y2**2 + y3**2 + y4**2

    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()

print(f"A = {A.item():.6f}")
print(f"B = {B.item():.6f}")
print(f"C = {C.item():.6f}")
print(f"D = {D.item():.6f}")
print(f"A + B = {(A + B).item():.6f} (mục tiêu 9)")
print(f"C - D = {(C - D).item():.6f} (mục tiêu 1)")
print(f"A + C = {(A + C).item():.6f} (mục tiêu 8)")
print(f"B - D = {(B - D).item():.6f} (mục tiêu 2)")
print(f"Tổng sai số bình phương = {sqerr.item():.12f}")
```

### Nhận xét và trả lời

Nghiệm thu được thỏa bốn phương trình với sai số số học rất nhỏ. Hệ có nhiều nghiệm vì bốn phương trình không độc lập: phương trình `B - D = 2` suy ra từ ba phương trình còn lại. Hệ chỉ có hạng 3 với 4 ẩn nên còn một bậc tự do. Họ nghiệm có thể viết là `A = t`, `B = 9 - t`, `C = 8 - t`, `D = 7 - t`.

## Bài 11 Bài tập áp dụng

```python
# 11.1 Tạo X có 24 phần tử và reshape
X = torch.arange(24)
print("X ban đầu:", X)
print("\nX reshape 4 x 6:\n", X.reshape(4, 6))
print("\nX reshape 2 x 12:\n", X.reshape(2, 12))
print("\nX reshape 2 x 3 x 4:\n", X.reshape(2, 3, 4))

# 11.2 Các phép toán với hai Tensor 3 x 4
A11 = torch.arange(1, 13, dtype=torch.float32).reshape(3, 4)
B11 = torch.arange(2, 14, dtype=torch.float32).reshape(3, 4)
print("\nA:\n", A11)
print("\nB:\n", B11)
print("\nA + B:\n", A11 + B11)
print("\nA * B:\n", A11 * B11)
print("\nA / B:\n", A11 / B11)
print("\nTrung bình theo cột của A:", A11.mean(dim=0))
print("Độ lệch chuẩn theo cột của A:", A11.std(dim=0))
print("Trung bình theo cột của B:", B11.mean(dim=0))
print("Độ lệch chuẩn theo cột của B:", B11.std(dim=0))
```

```python
# 11.3 Autograd với y = 2x^2 + 5x + 3 tại x = 2
x11 = torch.tensor(2.0, requires_grad=True)
y11 = 2 * x11**2 + 5 * x11 + 3
y11.backward()

manual_grad = 4 * x11.item() + 5
print("y(2) =", y11.item())
print("Gradient từ autograd =", x11.grad.item())
print("Gradient thủ công 4x + 5 =", manual_grad)
```

```python
# 11.4 Học hệ số của y = 2x^2 - 3x + 5
np.random.seed(SEED + 11)
torch.manual_seed(SEED + 11)

polynomial_11 = np.poly1d([2, -3, 5])
X_np = np.linspace(-3, 3, 50).reshape(-1, 1)
Y_np = polynomial_11(X_np)
XX_np = np.hstack([X_np**2, X_np, np.ones_like(X_np)])

x_train = torch.tensor(XX_np, dtype=torch.float32)
y_train = torch.tensor(Y_np, dtype=torch.float32)
w11 = torch.randn(3, 1, requires_grad=True)
optimizer11 = torch.optim.NAdam([w11], lr=0.01)

for epoch in range(4000):
    optimizer11.zero_grad()
    prediction = x_train @ w11
    loss11 = torch.mean((y_train - prediction) ** 2)
    loss11.backward()
    optimizer11.step()
    if epoch + 1 in (1, 100, 500, 1000, 2000, 4000):
        print(f"Epoch {epoch + 1:4d} - MSE = {loss11.item():.10f}")

learned11 = w11.detach().flatten()
target11 = torch.tensor([2.0, -3.0, 5.0])
print("Hệ số học được:", learned11.tolist())
print("Hệ số mục tiêu:", target11.tolist())
print("Sai lệch tuyệt đối:", torch.abs(learned11 - target11).tolist())
```

### Nhận xét và trả lời

- Tensor 24 phần tử có thể reshape thành `4 × 6`, `2 × 12` và `2 × 3 × 4` vì cả ba kích thước đều có tích bằng 24.
- Các phép `A+B`, `A*B`, `A/B` được thực hiện theo từng phần tử. Mean và std theo `dim=0` trả về kết quả cho từng cột.
- Với `y = 2x² + 5x + 3`, đạo hàm là `dy/dx = 4x + 5`; tại `x = 2`, gradient bằng `13`, trùng với autograd.
- Mô hình NAdam học lại các hệ số gần `[2, -3, 5]`, đồng thời MSE giảm về gần 0.
